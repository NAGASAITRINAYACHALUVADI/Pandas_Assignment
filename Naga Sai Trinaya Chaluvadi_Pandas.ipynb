## C – Q1. Broadcasting

````python
a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
````

**Shapes of a and b:**

Shape of A:  (3, 1) 

Shape of B: (1, 3)

**Predicted output / shape:**

[[11 12 13]

 [21 22 23]

 [31 32 33]]

(3, 3)

(3,3)

**How broadcasting works (and why it differs from adding two 1-D arrays):**

Arithmetic operations on NumPy arrays usually require compatible shapes. Broadcasting allows operations between arrays with different but compatible shapes by automatically expanding their dimensions.


**What happens if b = np.array([1, 2]) and why:**

[[11 12]
 [21 22]
 [31 32]]

 The shapes are compatible through broadcasting, so b is treated as (1, 2) and expanded across the rows of a, producing a (3, 2) result.

````python
# Your code to test the b = np.array([1, 2]) case (use try/except so the notebook doesn't stop)
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([1, 2])

try:
    result = a + b
    print(result)
    print(result.shape)
except ValueError as e:
    print("Error:", e)
````

## C – Q3. Outliers: mean vs median

````python
df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
````

**Predicted mean and median:**

Mean: 2522500.0
Median: 32500.0


**Effect of the extreme value on the mean:**


The extreme salary of 10,000,000 greatly increases the mean, making it much higher than the other salaries.

**Better statistic for a typical salary, and why:**

Median is better because it is less affected by extreme values and represents the typical salary more accurately.


````python
# Find the record with the highest salary
result = df.loc[df['Salary'].idxmax(), ['Employee', 'Salary']]
print(result)
````

**Why the record should be investigated before removing it:**

The record may be a genuine extreme value rather than an error. It should be verified before removing it to avoid losing valid data.

````python
df = df.drop_duplicates()
print(df)
````

**Explanation (one line):**

````python
df_ = df.copy()

df_['Marks'] = pd.to_numeric(df_['Marks'], errors='coerce')

print(df)
````

````python
print(df[df['Marks'].isna()])
````

### Step 6 – Identify marks outside 0–100 and decide how to handle them

````python
invalid_marks = df_[(df_['Marks'] < 0) | (df_['Marks'] > 100)]
print(invalid_marks)
````

````python
print("Cleaned DataFrame:")
print(df)

print("\nCleaning Report")
print("Records removed:", 15 - len(df))
print("Missing marks remaining:", df['Marks'].isna().sum())

invalid_marks = df[(df['Marks'] < 0) | (df['Marks'] > 100)]
print("Invalid marks identified:", len(invalid_marks))
````

**Explanation (one line):**

The code displays the cleaned DataFrame and reports the number of records removed, missing marks remaining, and marks outside the valid 0–100 range.
